```python
import pandas as pd
```

```python
CONF_DIR = 'conf'
DATA = 'data/sample.csv'
BASE_YAML_PATH = 'conf/base.yaml'
sim_data = pd.read_csv(DATA)
sim_data = sim_data.head(5)
sim_data
```

```python
import copy
def get_base_yaml(base_yaml_path):
    # print the yaml results in base_yaml_path  using PyYAML
    import yaml
    with open(base_yaml_path, 'r') as f:
        yaml_results = yaml.safe_load(f)
    
    return yaml_results

def generate_new_yaml(base_yaml, persona_name, persona_description):
    # generate a new yaml file from base_yaml
    # updated example_persona key to persona_name and persona_description
    new_yaml = copy.deepcopy(base_yaml)
    new_yaml['example_persona'] = persona_name + persona_description
    
    return new_yaml
    
    
def write_yaml(yaml_results, output_path):
    # write yaml_results to output_path
    import yaml
    with open(output_path, 'w') as f:
        yaml.dump(yaml_results, f)
    
    return output_path
    
    
def get_and_write_new_yaml(base_yaml_path, persona_name, persona_description, output_path):
    base_yaml = get_base_yaml(base_yaml_path)
    new_yaml = generate_new_yaml(base_yaml, persona_name, persona_description)
    output_path = write_yaml(new_yaml, output_path)
    
    return output_path
    

sim_data['yaml'] = sim_data.apply(lambda x: get_and_write_new_yaml(BASE_YAML_PATH, x['persona_name'], x['persona_description'], CONF_DIR + '/' + x['name'].lower().replace(" ", "_") + '.yaml'), axis=1)
sim_data['yaml']
```
