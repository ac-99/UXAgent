```python
from tqdm import tqdm
tqdm.pandas()
def run_sim(persona_name, persona_description, flow_name, flow_description,starting_url):
    import subprocess
    # config_name = Path(yaml_path).stem  # "conf/sim_73.yaml" -> "sim_73"
    # print the command that will be run
    command = ['python', '-m', 'src.simulated_web_agent.main', '--persona', persona_name +" "+ persona_description, '--intent', flow_name +" "+ flow_description, '--start-url', starting_url, '--record']
    if SAMPLE:
        command.append('--max-steps')
        command.append(str(2))
    print(command)
    subprocess.run(command)
    
sim_data['run'] = sim_data.progress_apply(lambda x: run_sim(x['persona_name'], x['persona_description'], x['flow_name'], x['flow_description'],x['starting_url']), axis=1)    
```
