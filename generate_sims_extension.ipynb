```python
from tqdm import tqdm                                                                                                    
import subprocess                                                                                                        
tqdm.pandas()                                                                                                            
                                                                                                                        
TIMEOUT = 40 * 60  # 40 minutes in seconds

def run_sim(persona_name, persona_description, flow_name, flow_description, starting_url):                               
    command = ['python', '-m', 'src.simulated_web_agent.main', '--persona', persona_name + " " + persona_description,
'--intent', flow_name + " " + flow_description, '--start-url', starting_url, '--record']                                 
    if SAMPLE:  
        command.extend(['--max-steps', str(2)])                                                                          
    else:       
        command.extend(['--max-steps', str(MAX_STEPS)])
    print(command)                                                                                                       
    try:
        subprocess.run(command, timeout=TIMEOUT)                                                                         
    except subprocess.TimeoutExpired:
        print(f"TIMEOUT: {starting_url} - {flow_name}")
                                                                                                                        
sim_data['run'] = sim_data.progress_apply(lambda x: run_sim(x['persona_name'], x['persona_description'], x['flow_name'], 
x['flow_description'], x['starting_url']), axis=1)  
```
